```python
import random

def get_conflicts(state):
    """Calculates the number of conflicting pairs of queens on the board."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Check if they are in the same row or on the same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    """Generates neighbors by moving one queen in its column and returns the best state."""
    n = len(state)
    current_conflicts = get_conflicts(state)
    best_state = list(state)
    best_conflicts = current_conflicts

    # Examine all possible single-queen moves
    for col in range(n):
        original_row = state[col]
        for row in range(n):
            if row == original_row:
                continue

            # Create a neighbor state by moving queen in 'col' to 'row'
            neighbor = list(state)
            neighbor[col] = row
            neighbor_conflicts = get_conflicts(neighbor)

            if neighbor_conflicts < best_conflicts:
                best_conflicts = neighbor_conflicts
                best_state = neighbor

    return tuple(best_state), best_conflicts

def hill_climbing_n_queens(n=4, max_restarts=100):
    """Solves N-Queens using Hill Climbing with Random Restarts."""
    for restart in range(max_restarts):
        # Start with a random state (one queen per column)
        current_state = tuple(random.randint(0, n - 1) for _ in range(n))
        current_conflicts = get_conflicts(current_state)

        while True:
            neighbor, neighbor_conflicts = get_best_neighbor(current_state)

            # If we found a global minimum (solution)
            if neighbor_conflicts == 0:
                return neighbor, restart

            # If the best neighbor does not improve the current state, we are at a local minimum
            if neighbor_conflicts >= current_conflicts:
                break

            current_state = neighbor
            current_conflicts = neighbor_conflicts

    return None, max_restarts

def print_board(state):
    """Helper to print the chessboard nicely."""
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

if __name__ == "__main__":
    n = 4
    print(f"--- Solving {n}-Queens using Hill Climbing ---")
    solution, restarts = hill_climbing_n_queens(n)

    if solution:
        print(f"Solution found after {restarts} random restart(s):")
        print(f"State Representation (rows per column): {solution}\n")
        print_board(solution)
    else:
        print("Failed to find a solution within the restart limit.")
```

```python
import random

def get_conflicts(state):
    """Calculates the number of conflicting pairs of queens on the board."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Check if they are in the same row or on the same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    """Generates neighbors by moving one queen in its column and returns the best state."""
    n = len(state)
    current_conflicts = get_conflicts(state)
    best_state = list(state)
    best_conflicts = current_conflicts

    # Examine all possible single-queen moves
    for col in range(n):
        original_row = state[col]
        for row in range(n):
            if row == original_row:
                continue

            # Create a neighbor state by moving queen in 'col' to 'row'
            neighbor = list(state)
            neighbor[col] = row
            neighbor_conflicts = get_conflicts(neighbor)

            if neighbor_conflicts < best_conflicts:
                best_conflicts = neighbor_conflicts
                best_state = neighbor

    return tuple(best_state), best_conflicts

def hill_climbing_n_queens_from_start(initial_state, max_restarts=100):
    """Solves N-Queens starting from a user-defined initial state, falling back to random restarts if stuck."""
    n = len(initial_state)
    current_state = initial_state
    current_conflicts = get_conflicts(current_state)

    if current_conflicts == 0:
        return current_state, 0

    # First attempt: climb from the user's initial state
    while True:
        neighbor, neighbor_conflicts = get_best_neighbor(current_state)
        if neighbor_conflicts == 0:
            return neighbor, 0
        if neighbor_conflicts >= current_conflicts:
            break
        current_state = neighbor
        current_conflicts = neighbor_conflicts

    # If stuck, fallback to random restarts
    print("Local minimum reached from initial placement. Starting random restarts...")
    for restart in range(1, max_restarts + 1):
        current_state = tuple(random.randint(0, n - 1) for _ in range(n))
        current_conflicts = get_conflicts(current_state)

        while True:
            neighbor, neighbor_conflicts = get_best_neighbor(current_state)
            if neighbor_conflicts == 0:
                return neighbor, restart
            if neighbor_conflicts >= current_conflicts:
                break
            current_state = neighbor
            current_conflicts = neighbor_conflicts

    return None, max_restarts

def print_board(state):
    """Helper to print the chessboard nicely."""
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

if __name__ == "__main__":
    try:
        n_input = input("Enter board size N (e.g., 4): ")
        n = int(n_input)

        print(f"For each column (0 to {n-1}), enter the starting row index (0 to {n-1}).")
        initial_list = []
        for col in range(n):
            row_val = int(input(f"  Row for column {col}: "))
            if not (0 <= row_val < n):
                raise ValueError(f"Row index must be between 0 and {n-1}")
            initial_list.append(row_val)

        initial_state = tuple(initial_list)
        print(f"\nInitial user configuration: {initial_state}")
        print_board(initial_state)
        print(f"Initial conflicts: {get_conflicts(initial_state)}\n")

        solution, restarts = hill_climbing_n_queens_from_start(initial_state)

        if solution:
            print(f"Solution found after {restarts} random restart(s):")
            print(f"State Representation (rows per column): {solution}\n")
            print_board(solution)
        else:
            print("Failed to find a solution within the restart limit.")
    except ValueError as e:
        print(f"Invalid input: {e}")
```
